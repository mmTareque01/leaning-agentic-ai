# Learning Agentic AI from scratch
Here we will be learning the basic prompting on any AI

```python
# Importing api key from env file & creating groq client instance

import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("GROQ_API_KEY")

client = Groq(api_key=api_key)
```

```python
def askGroq(prompt):
  completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": prompt
      }
    ],
)
  return completion
```

```python
response = askGroq("Suggest me the best path of learning agentic AI")

print(response.choices[0].message.content)
```
