```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("GROQ_API_KEY")

client = Groq(api_key=api_key)
```

```python
# Version 1
def askGroq(prompt):
  completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": prompt
      }
    ],
)
  return completion
```

```python
response = askGroq("Suggest me the best path of learning agentic AI")

print(response.choices[0].message.content)
```

```python
# version 2
def askGroq(prompt):
  completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": prompt
      }
    ],
    temperature=0.2,
    max_completion_tokens=1500,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)
  return completion
```

```python
myAns = askGroq('Suggest me the best path of learning agenctic AI')
# print(myAns)
for chunk in myAns:
    print(chunk.choices[0].delta.content or "", end="")
```

```python
def create_plan(task):

    prompt = f"""
    You are an AI planning agent.
    Break the task into clear steps.

    Task: {task}

    Return only the steps.
    """

    response = askGroq(prompt)

    return response
```

```python
# version 3
def askGroq(prompt):
  completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": prompt
      }
    ],
    temperature=0.2,
    max_completion_tokens=1500,
    top_p=1,
    reasoning_effort="medium",
    # stream=True,
    stop=None
)
  return completion
```

```python
def create_plan(task):

    prompt = f"""
You are an AI planning agent.

Break the task into steps.

Return ONLY valid JSON in this format:

{{
 "task": "{task}",
 "steps": [
   "step1",
   "step2",
   "step3"
 ]
}}

Task: {task}
"""

    return askGroq(prompt)
```

```python
import json

response = create_plan("Build a weather web app")

```

```python
content = response.choices[0].message.content

data = json.loads(content)

print(data["task"])
print(data["steps"])
```

```python
for step in data["steps"]:
    print("Executing:", step)
```

```python
def save_to_file(filename, content):
    with open(filename, "w") as f:
        f.write(content)

    return "File saved successfully"

def read_file(filename):
    with open(filename, "r") as f:
        return f.read()

tools = {
    "save_file": save_to_file,
    "read_file": read_file
}
```

```python
for step in data["steps"]:
    print("\nExecuting:", step)

    if "write" in step.lower() or "save" in step.lower():
        result = tools["save_file"]("output.txt", step)
        print("Tool Result:", result)

    else:
        print("No tool needed")
```
