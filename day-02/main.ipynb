```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("GROQ_API_KEY")

client = Groq(api_key=api_key)
```

```python
# version 2
def askGroq(prompt):
  completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": prompt
      }
    ],
    temperature=0.2,
    max_completion_tokens=1500,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)
  return completion
```

```python
myAns = askGroq('Suggest me the best path of learning agenctic AI')
# print(myAns)
for chunk in myAns:
    print(chunk.choices[0].delta.content or "", end="")
```

```python
def create_plan(task):

    prompt = f"""
    You are an AI planning agent.
    Break the task into clear steps.

    Task: {task}

    Return only the steps.
    """

    response = askGroq(prompt)

    return response
```

```python
plan = create_plan("Build a weather web app")
for chunk in plan:
    print(chunk.choices[0].delta.content or "", end="")
```
